### Running games

```python
from Game import Game
from Agents.RuleAgent import RuleAgent
from Structures.PlayerID import PlayerID
from Structures.ValidPlayerColors import ValidPlayerColors
from Parameters import Parameters
from AutomatedPlaytest import playtestNtimes
from CalculateCriteria import run
import os

root = "/home/lana/Documentos/Risk-Generation/"
```

#### Running a single game

```python
showActions = False
mapPath = root + "parameters/map1.json"
troopsWonBeginTurn = "3"
defenseDices = "2"
initialTerritoriesMode = "random"
troopsToNewTerritory = "min"

parameters = Parameters(mapPath, troopsWonBeginTurn, defenseDices, initialTerritoriesMode, troopsToNewTerritory)

agent1 = RuleAgent(PlayerID("Player1", ValidPlayerColors.BLUE))
agent2 = RuleAgent(PlayerID("Player2", ValidPlayerColors.RED))
agent3 = RuleAgent(PlayerID("Player3", ValidPlayerColors.GREEN))
agent4 = RuleAgent(PlayerID("Player4", ValidPlayerColors.PURPLE))

game = Game(showActions=showActions, parameters=parameters, listOfPlayers=[agent1, agent2, agent3, agent4])

metrics = game.playtest()

metricsFile = root + "parameters/game" + troopsWonBeginTurn + "-" + defenseDices + "-" + initialTerritoriesMode + "-" + troopsToNewTerritory + ".txt"
metrics.appendToFile(metricsFile, len(game.listOfPlayers))
```

#### Running multiple games

```python
mapPath = "parameters/map1.json"
troopsWonBeginTurn = "3"
defenseDices = "2"
initialTerritoriesMode = "random"
troopsToNewTerritory = "min"

game = Parameters(mapPath, troopsWonBeginTurn, defenseDices, initialTerritoriesMode, troopsToNewTerritory)
playtestNtimes(game)
filename = "parameters/game" + troopsWonBeginTurn + "-" + defenseDices + "-" + initialTerritoriesMode + "-" + troopsToNewTerritory + ".txt"
run(filename)
os.remove(filename)
```
